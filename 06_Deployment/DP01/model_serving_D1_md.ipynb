### ✅ 체크포인트

다음 질문에 답할 수 있다면, 이 섹션의 학습 목표를 달성한 것입니다:

1. `state_dict`로 저장한 `.pth` 파일을 불러올 때, 모델 클래스 정의가 필요한 이유는 무엇입니까?  
: pth파일에는 가중치(숫자 값)만 저장되고, 모델의 구조(어떤 레이어가 어떤 순서로 연결되는지)는 저장되지 않는다. 따라서 가중치를 제자리에 입히기 위해 원본과 똑같은 모델 클래스 뼈대를 먼저 선언해야한다.  

2. `torch.jit.trace`와 `torch.jit.script`는 각각 어떤 상황에서 사용합니까?  
- torch.jit.trace :  모델 내부에 if/else 분기 같은 동적 제어 흐름이 없을 때 사용. 더미 입력을 흘려보내 경로를 기록하는 방식이라 더 안정적.  
- torch.jit.script : 입력 데이터에 따라 연산 경로가 달라지는 동적인 로직(if/else, for 루프 등)이 포함된 모델일 때 사용.  

3. ONNX의 가장 큰 장점은 무엇이며, 어떤 상황에서 선택해야 합니까?  
- 가장 큰 장점: 프레임워크 및 언어에 구애받지 않는 호환성(독립성)입니다. PyTorch로 만든 모델을 TensorFlow에서 돌릴 수도 있고, Python 없이 C++나 Java 환경에서도 실행할 수 있습니다.
- 선택 상황: 다른 프레임워크나 하드웨어 최적화 엔진(TensorRT 등)을 사용해야 할 때, 혹은 매우 빠른 추론 속도(성능 최적화)가 중요한 프로덕션 환경에 배포할 때 주로 선택합니다.  

4. `torch.onnx.export()`에서 `dynamic_axes`를 지정하지 않으면 어떤 문제가 생길 수 있습니까?  
: 입력 데이터의 축 크기(예: batch_size)가 학습할 때 사용한 크기(예: 1)로 완전히 고정되어 버리는 문제가 발생합니다. 실제 서비스 환경에서는 동시에 2명, 3명의 요청이 들어와 배치 크기가 가변적으로 변할 수 있는데, 이를 지정하지 않으면 고정된 크기 외의 요청이 들어왔을 때 모델이 처리하지 못해 에러가 납니다.
