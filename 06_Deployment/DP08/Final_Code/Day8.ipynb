## 패키지 설치

PowerShell 터미널에서 아래 명령을 실행합니다. %pip는 노트북 셀에서만 사용하는 문법이므로 터미널에 입력하지 않습니다.

python -m pip install -r requirements.txt


```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
required_files = [
    'app/auth.py',
    'app/schemas.py',
    'app/model_service.py',
    'app/main.py',
    'frontend/app.py',
]

for relative_path in required_files:
    path = PROJECT_ROOT / relative_path
    print(('OK  ' if path.exists() else '없음 ') + relative_path)
```

## 구현 파일 확인

아래 셀은 Day 8에서 작성한 파일 내용을 노트북에서 보여 줍니다. 필요한 경우 VS Code에서 같은 파일을 열어 수정합니다.


```python
from IPython.display import Markdown, display

for relative_path in required_files:
    path = PROJECT_ROOT / relative_path
    if path.exists():
        display(Markdown('## ' + relative_path))
        print(path.read_text(encoding='utf-8'))
```

## 실행

VS Code PowerShell 터미널 1: uvicorn app.main:app --reload

VS Code PowerShell 터미널 2: streamlit run frontend/app.py

Streamlit에서는 test-key-001 또는 test-key-002를 입력합니다. FastAPI 문서는 http://127.0.0.1:8000/docs에서 확인합니다.


## 오류 처리 확인

- API Key 미입력 또는 wrong-key: 401
- 10자 미만 텍스트: 422
- 512 토큰 초과 입력: 422
- 동시에 요청: 429
- 모델 로드 실패: 503
- FastAPI 서버 미실행: Streamlit 연결 오류
